```python
primes = [2, 3, 5, 7]
primes.append(11)
primes.append(13)
print('After append:', primes)
primes.remove(3)
print('After remove:', primes)
primes.sort()
print('Sorted      :', primes)
```

```python
data = [4, 8, 15, 16, 23, 42]
total  = sum(data)
count  = len(data)
mean   = total / count
minval = min(data)
maxval = max(data)
print(f'Data  : {data}')
print(f'Sum   : {total}')
print(f'Count : {count}')
print(f'Mean  : {mean:.2f}')
print(f'Min   : {minval}')
print(f'Max   : {maxval}')
```

```python
point = (3, 4)
print('Point  :', point)
print('X-coord:', point[0])
print('Y-coord:', point[1])
distance = (point[0]**2 + point[1]**2) ** 0.5
print(f'Distance from origin: {distance:.4f}')
```
